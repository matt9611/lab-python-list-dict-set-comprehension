# Lab | List, Dict and Set Comprehension — Solved

This notebook completes the customer-orders exercise using comprehensions.

## Solution

The program initializes inventory, collects customer orders, calculates order statistics and total price, and removes products whose inventory reaches zero.

```python
def initialize_inventory(products):
    """Create an inventory dictionary with one quantity per product."""
    return {
        product: int(input(f"Enter the quantity of {product}s available: "))
        for product in products
    }


def get_customer_orders():
    """Ask for the number of orders and collect each product name."""
    number_of_orders = int(input("Enter the number of customer orders: "))
    return [
        input("Enter the name of a product that a customer wants to order: ")
        for _ in range(number_of_orders)
    ]


def calculate_order_statistics(customer_orders, products):
    """Return order count and the percentage of unique products ordered."""
    total_products_ordered = len(customer_orders)
    unique_products_ordered = set(customer_orders)
    percentage_unique = (
        len(unique_products_ordered) / len(products) * 100
        if products else 0
    )
    return total_products_ordered, percentage_unique


def calculate_total_price(customer_orders):
    """Ask for each product price and sum the prices with a comprehension."""
    prices = {
        product: float(input(f"Enter the price of {product}: "))
        for product in customer_orders
    }
    return sum(prices.values())


def update_inventory(inventory, customer_orders):
    """Subtract one unit per order and filter out zero-quantity products."""
    updated_inventory = {
        product: quantity - 1 if product in customer_orders else quantity
        for product, quantity in inventory.items()
        if quantity - (1 if product in customer_orders else 0) > 0
    }
    return updated_inventory
```

```python
# Run the complete program
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = initialize_inventory(products)
customer_orders = get_customer_orders()

total_products_ordered, percentage_unique = calculate_order_statistics(
    customer_orders, products
)

print("\nOrder Statistics:")
print(f"Total Products Ordered: {total_products_ordered}")
print(f"Percentage of Unique Products Ordered: {percentage_unique:.1f}")

updated_inventory = update_inventory(inventory, customer_orders)
print("\nUpdated Inventory:")
for product, quantity in updated_inventory.items():
    print(f"{product}: {quantity}")

total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price}")
```

## Notes

- The inventory is created with a dictionary comprehension.
- Orders are collected with a list comprehension.
- Unique products are identified with a set.
- Prices are collected with a dictionary comprehension and summed.
- The updated inventory uses a dictionary comprehension with a filter so products at zero quantity are removed.
